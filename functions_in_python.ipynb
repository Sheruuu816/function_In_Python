```python
# 1. Even or Odd Function

def is_even(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

num = int(input("Enter a number: "))

print(is_even(num))
```

```python
# 2. Largest of Two Numbers

def largest(a, b):
    if a > b:
        return a
    else:
        return b

n1 = int(input("Enter 1st number: "))
n2 = int(input("Enter 2nd number: "))

print("Largest : ", largest(n1, n2))
```

```python
# 3. Print Numbers

def print_numbers(n):
    for i in range(1, n + 1):
        print(i)

num = int(input("Enter a number: "))

print_numbers(num)
```

```python
# 4. Sum of Numbers

def sum_n(n):
    sum = 0

    for i in range(1, n + 1):
        sum = sum + i

    return sum

num = int(input("Enter a number: "))

print("Sum : ", sum_n(num))
```

```python
# 5. Multiplication Table

def table(n):
    for i in range(1, 11):
        print(n, "x", i, "=", n * i)

num = int(input("Enter a number: "))

table(num)
```

```python
# 6. Count Even Numbers

def count_even(numbers):
    count = 0

    for i in numbers:
        if i % 2 == 0:
            count = count + 1

    return count

numbers = [1, 2, 3, 4, 6]

print("Even numbers : ", count_even(numbers))
```

```python
# 7. Find Maximum

def find_max(numbers):
    largest = numbers[0]

    for i in numbers:
        if i > largest:
            largest = i

    return largest

numbers = [10, 25, 5, 40, 15]

print("Maximum : ", find_max(numbers))
```

```python
# 11. FizzBuzz

def fizzbuzz():
    for i in range(1, 101):

        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

fizzbuzz()
```

```python
# 12. Palindrome Checker

def palindrome(text):
    if text == text[::-1]:
        return True
    else:
        return False

text = input("Enter a string: ")
print(palindrome(text))
```

```python
# 13. Count Vowels

def count_vowels(text):
    count = 0

    for i in text:
        if i == "a" or i == "e" or i == "i" or i == "o" or i == "u":
            count = count + 1

    return count

text = input("Enter a string: ")

print("Number of vowels : ", count_vowels(text))
```

```python
# 14. Second Largest Number

def second_largest(numbers):
    largest = numbers[0]
    second = numbers[0]

    for i in numbers:
        if i > largest:
            second = largest
            largest = i
        elif i > second and i != largest:
            second = i

    return second

numbers = [10, 25, 5, 40, 15]

print("Second largest : ", second_largest(numbers))
```

```python
# 15. Guessing Game

import random

def guessing_game():
    number = random.randint(1, 10)

    while True:
        guess = int(input("Guess a number between 1 and 10: "))

        if guess == number:
            print("Correct guess!")
            break
        elif guess < number:
            print("Try a greater number")
        else:
            print("Try a smaller number")

guessing_game()
```

```python
# 16. Grade Function

def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"

marks = int(input("Enter marks: "))

print("Grade : ", grade(marks))
```
